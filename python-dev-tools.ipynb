### 3.0 Linting Tool Taxonomy *(~15 min)*

**Context.** New developer conflates tools with different purposes.

**Task.** Categorize the following tools into: Formatters, Linters, Type Checkers:
`black`, `ruff`, `flake8`, `pylint`, `mypy`, `pyright`, `isort`, `ruff format`

Also answer: Is mypy a PEP 8 checker? What is the recommended modern stack?

## Answer

Formatters (auto-rewrite your code to enforce style):
black, isort, ruff format
Linters (analyze code and report problems, but don't auto-rewrite):
flake8, pylint, ruff
Type Checkers (verify type annotation correctness):
mypy, pyright

Is mypy a PEP 8 checker? No. mypy is purely a type checker — it verifies that your type annotations are consistent and correct. It doesn't care about naming conventions, line length, whitespace, or any of the stylistic rules PEP 8 covers
